importing necessary libraries and the data set

```python
import numpy as np
import pandas as pd
import plotly.io as pio
pio.renderers.default = "iframe"
import plotly.express as px
from textblob import TextBlob
import plotly.graph_objects as go
```

```python
# starting our journey with india and USA data
import json
# Load the datasets
# Use forward slashes and no backslashes before the space
# Define the folder path to save typing it over and over
folder_path = '/Users/abhimanyuchettiar/Desktop/youtube dataset/'
df_in = pd.read_csv('/Users/abhimanyuchettiar/Desktop/youtube dataset/INvideos.csv')
df_us = pd.read_csv('/Users/abhimanyuchettiar/Desktop/youtube dataset/USvideos.csv')

# Verify it loaded
print(f"India data loaded: {len(df_in)} rows")
print(f"US data loaded: {len(df_us)} rows")
# Function to extract category names from JSON
def get_category_mapping(json_file):
    with open(json_file, 'r') as f:
        data = json.load(f)
    
    mapping = {}
    for category in data['items']:
        mapping[int(category['id'])] = category['snippet']['title']
    return mapping

# Create the mapping dictionaries
in_categories = get_category_mapping(folder_path +'IN_category_id.json')
us_categories = get_category_mapping(folder_path + 'US_category_id.json')

# Map the IDs to actual names in our DataFrames
df_in['category_name'] = df_in['category_id'].map(in_categories)
df_us['category_name'] = df_us['category_id'].map(us_categories)

# Quick check
print(df_in[['title', 'category_name']].head())
```

Cleaning the dataset

```python
# Convert trending_date: YY.DD.MM -> Datetime
df_in['trending_date'] = pd.to_datetime(df_in['trending_date'], format='%y.%d.%m')
df_us['trending_date'] = pd.to_datetime(df_us['trending_date'], format='%y.%d.%m')

# Convert publish_time to Datetime
df_in['publish_time'] = pd.to_datetime(df_in['publish_time'])
df_us['publish_time'] = pd.to_datetime(df_us['publish_time'])

# Deduplication: Keep the version with the most views
df_in = df_in.sort_values('views', ascending=False).drop_duplicates('video_id')
df_us = df_us.sort_values('views', ascending=False).drop_duplicates('video_id')

# New Column: Days taken to trend
df_in['days_to_trend'] = (df_in['trending_date'] - df_in['publish_time'].dt.tz_localize(None)).dt.days
df_us['days_to_trend'] = (df_us['trending_date'] - df_us['publish_time'].dt.tz_localize(None)).dt.days
```

Data Analysis

```python
Category Showdown (India vs US), Lets see which genre dominate the trending page
```

```python
# Top categories for India
top_categories_in = df_in['category_name'].value_counts().reset_index()
top_categories_in.columns = ['Category', 'Count']

# Top categories for US
top_categories_us = df_us['category_name'].value_counts().reset_index()
top_categories_us.columns = ['Category', 'Count']

# Plotting Side-by-Side
fig_in = px.bar(top_categories_in.head(10), x='Count', y='Category', orientation='h', 
             title='Top 10 Trending Categories in India', color='Count')
fig_us = px.bar(top_categories_us.head(10), x='Count', y='Category', orientation='h', 
             title='Top 10 Trending Categories in US', color='Count')

fig_in.show()
fig_us.show()
```

The viral speed analysis
how long it takes for a video to trend

```python
# Distribution of days taken to reach trending page
fig_viral = px.histogram(df_in, x='days_to_trend', nbins=50, 
                        title='Days Taken to Reach Trending (India)',
                        labels={'days_to_trend': 'Days from Publish to Trending'},
                        range_x = [0,15]) 

fig_viral.show()
```

Engagement Correlation (Heatmap)
lets see which metrics are most tied to views. Does a high comment count correlate with views more than likes do?

```python
# 1. Calculate the correlation matrix
columns_to_study = ['views', 'likes', 'dislikes', 'comment_count']
corr_matrix = df_us[columns_to_study].corr()

# 2. Create the interactive heatmap
fig = px.imshow(
    corr_matrix,
    text_auto='.2f',                # Adds the annotations (replaces annot=True)
    aspect="auto",
    color_continuous_scale='RdBu_r', # Reverses Red-Blue to match Seaborn style
    labels=dict(color="Correlation"),
    title='Engagement Correlation Heatmap (US)',
    zmin=-1, zmax=1                  # Ensures scale stays within correlation limits
)

# 3. Adjust layout for readability
fig.update_layout(
    xaxis_title="Metrics",
    yaxis_title="Metrics"
)

fig.show()
```

When is the best time to publish ?

```python
# Extract day and hour
df_in['publish_day'] = df_in['publish_time'].dt.day_name()
df_in['publish_hour'] = df_in['publish_time'].dt.hour

# Count uploads per day
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df_day = df_in['publish_day'].value_counts().reindex(day_order).reset_index()

fig_day = px.bar(df_day, x='publish_day', y='count', 
                 title='Best Day to Publish for Trending (India)',
                 labels={'publish_day': 'Day of Week', 'count': 'Number of Trending Videos'})
fig_day.show()
```

Like to View Ratio (Efficiency Metric)

```python
# Create the ratio column
df_in['like_ratio'] = (df_in['likes'] / df_in['views']) * 100

# Average like ratio per category
cat_ratio = df_in.groupby('category_name')['like_ratio'].mean().sort_values(ascending=False).reset_index()

fig_ratio = px.bar(cat_ratio, x='like_ratio', y='category_name', orientation='h',
                   title='Average Like-to-View Ratio per Category (India)',
                   labels={'like_ratio': 'Likes per 100 Views'})
fig_ratio.show()
```

Title Analysis (Do capitalized titles perform better than regular ones?)

```python
def contains_capitalized_word(title):
    for word in title.split():
        if word.isupper() and len(word) > 1 and word.isalpha():
            return True
    return False

df_in['has_caps'] = df_in['title'].apply(contains_capitalized_word)

# Compare average views
caps_comparison = df_in.groupby('has_caps')['views'].mean().reset_index()

fig_caps = px.pie(caps_comparison, values='views', names='has_caps', 
                 title='Average Views: Titles with ALL CAPS vs. Regular',
                 labels={'has_caps': 'Contains All-Caps Word'})
fig_caps.show()
```
