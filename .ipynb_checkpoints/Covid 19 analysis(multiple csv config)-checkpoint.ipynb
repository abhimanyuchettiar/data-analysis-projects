importing necessary libraries and the dataset

```python
import numpy as np
import pandas as pd
import plotly.io as pio
pio.renderers.default = "iframe"
import plotly.express as px
from textblob import TextBlob
import plotly.graph_objects as go
```

```python
# No backslashes before spaces or parentheses
df_1 = pd.read_csv(r"/Users/abhimanyuchettiar/Downloads/archive (6)/covid_19_india.csv")
df_2 = pd.read_csv(r"/Users/abhimanyuchettiar/Downloads/archive (6)/StatewiseTestingDetails.csv")
```

Data cleaning

```python
# Convert Date columns
df_1['Date'] = pd.to_datetime(df_1['Date'])
df_2['Date'] = pd.to_datetime(df_2['Date'])

# Check the change
print(df_1['Date'].dtype) # Should be datetime64[ns]
```

```python
#checking for discrepencies
states_1 = set(df_1['State/UnionTerritory'].unique())
states_2 = set(df_2['State'].unique())

print("States in df_1 but not df_2:", states_1 - states_2)
print("States in df_2 but not df_1:", states_2 - states_1)
```

```python
# Check for missing values
print(df_1.isnull().sum())

# Drop columns that are mostly empty or redundant
# which were only used in the first few weeks.
df_1.drop(['ConfirmedIndianNational', 'ConfirmedForeignNational'], axis=1, inplace=True, errors='ignore')
```

```python
# Calculate daily new cases per state
df_1 = df_1.sort_values(by=['State/UnionTerritory', 'Date'])
df_1['New_Cases'] = df_1.groupby('State/UnionTerritory')['Confirmed'].diff().fillna(0)
```

```python
# Create a mapping dictionary for the messy names
state_fix = {
    'Karanataka': 'Karnataka',
    'Maharashtra***': 'Maharashtra',
    'Madhya Pradesh***': 'Madhya Pradesh',
    'Bihar****': 'Bihar',
    'Himanchal Pradesh': 'Himachal Pradesh',
    'Telengana': 'Telangana',
    'Dadra and Nagar Haveli': 'Dadra and Nagar Haveli and Daman and Diu',
    'Daman & Diu': 'Dadra and Nagar Haveli and Daman and Diu'
}

# Apply the fix to df_1
df_1['State/UnionTerritory'] = df_1['State/UnionTerritory'].replace(state_fix)

# Drop rows that don't represent actual states (Unassigned, etc.)
drop_labels = ['Cases being reassigned to states', 'Unassigned']
df_1 = df_1[~df_1['State/UnionTerritory'].isin(drop_labels)]

print("Remaining discrepancies:", set(df_1['State/UnionTerritory'].unique()) - set(df_2['State'].unique()))
```

Data analysis

```python
# Rename columns to match for the merge
df_2 = df_2.rename(columns={'State': 'State/UnionTerritory'})

# Merge df_1 and df_2 on Date and State
df_merged = pd.merge(df_1, df_2[['Date', 'State/UnionTerritory', 'TotalSamples']], 
                     on=['Date', 'State/UnionTerritory'], 
                     how='left')

# Forward fill missing 'TotalSamples' 
# (States didn't report testing every single day, so we carry over the last known value)
df_merged['TotalSamples'] = df_merged.groupby('State/UnionTerritory')['TotalSamples'].ffill().fillna(0)
```

```python
# Calculate Daily New Samples (Testing)
df_merged['Daily_Tests'] = df_merged.groupby('State/UnionTerritory')['TotalSamples'].diff().fillna(0)

# Calculate Positivity Rate (New Cases / Daily Tests)
# We use a small epsilon (1e-6) to avoid division by zero errors
df_merged['Positivity_Rate'] = (df_merged['New_Cases'] / (df_merged['Daily_Tests'] + 1e-6)) * 100

# Cap the rate at 100% (sometimes data anomalies show > 100%)
df_merged['Positivity_Rate'] = df_merged['Positivity_Rate'].clip(0, 100)
```

```python
print(df_merged.groupby('State/UnionTerritory')['Positivity_Rate'].mean().sort_values(ascending=False).head(10))
```

```python
df_merged['Rolling_Positivity'] = df_merged.groupby('State/UnionTerritory')['Positivity_Rate'].transform(lambda x: x.rolling(window=7).mean())
```

Cumulative case growth

```python
import plotly.express as px

# Aggregating data for the whole of India
df_india = df_merged.groupby('Date')[['Confirmed', 'Cured', 'Deaths']].sum().reset_index()

fig1 = px.line(df_india, x="Date", y=["Confirmed", "Cured", "Deaths"], 
              title="Overall COVID-19 Growth in India",
              labels={"value": "Number of Cases", "variable": "Category"},
              color_discrete_map={"Confirmed": "blue", "Cured": "green", "Deaths": "red"})
fig1.show()
```

Top 10 most affected states

```python
# Getting the latest cumulative data for each state
latest_state_data = df_merged.groupby('State/UnionTerritory').max().sort_values(by='Confirmed', ascending=False).reset_index()

fig2 = px.bar(latest_state_data.head(10), x='State/UnionTerritory', y='Confirmed',
             title="Top 10 States by Confirmed Cases",
             color='Confirmed', color_continuous_scale='Reds')
fig2.show()
```

State wise recovery vs mortality

```python
latest_state_data['Recovery_Rate'] = (latest_state_data['Cured'] / latest_state_data['Confirmed']) * 100
latest_state_data['Fatality_Rate'] = (latest_state_data['Deaths'] / latest_state_data['Confirmed']) * 100

fig3 = px.scatter(latest_state_data, x='Recovery_Rate', y='Fatality_Rate', 
                 text='State/UnionTerritory', size='Confirmed', color='State/UnionTerritory',
                 title="Recovery vs. Fatality Rate by State")
fig3.update_traces(textposition='top center')
fig3.show()
```

Daily New cases(area chart)

```python
# Aggregate state-level daily data to get National daily data
df_india_daily = df_merged.groupby('Date')[['New_Cases', 'Daily_Tests', 'Cured', 'Deaths']].sum().reset_index()

# Smooth the data with a 7-day rolling average to remove weekend reporting lags
df_india_daily['7_Day_MA'] = df_india_daily['New_Cases'].rolling(window=7).mean()

print(df_india_daily.tail())
```

```python
fig4 = px.area(df_india_daily, x="Date", y="New_Cases", 
              title="Daily New Cases in India (The Waves)",
              color_discrete_sequence=['orange'])
fig4.show()
```

Testing vs Positivity Rate 

```python
# 1. Aggregate ALL relevant columns from the merged dataframe
df_india = df_merged.groupby('Date')[['Confirmed', 'Cured', 'Deaths', 'New_Cases', 'Daily_Tests']].sum().reset_index()

# 2. Now run the smoothing calculation
df_india['Rolling_Positivity'] = (
    df_india['New_Cases'].rolling(window=7).mean() / 
    (df_india['Daily_Tests'].rolling(window=7).mean() + 1e-6)
) * 100

# 3. Plotting with Plotly
import plotly.express as px

fig5 = px.line(df_india, x='Date', y='Rolling_Positivity', 
              title="7-Day Rolling Average of COVID-19 Positivity Rate in India",
              labels={'Rolling_Positivity': 'Positivity Rate (%)'},
              template="plotly_dark")

fig5.show()
```

Geographic Spread

```python
import requests
import json

# Fetching the GeoJSON for Indian States
repo_url = 'https://gist.githubusercontent.com/jbrobst/56c13bbbf9d97d187fea01ca62ea5112/raw/e388c4cae20aa53cb5090210a42ebb9b765c0a36/india_states.geojson'
india_states = requests.get(repo_url).json()

print(india_states['features'][0]['properties'])
```

```python
# Create a summary of the latest total cases per state
latest_map_data = df_merged.groupby('State/UnionTerritory')['Confirmed'].max().reset_index()
```

```python
fig6 = px.choropleth(
    latest_map_data,
    geojson=india_states,
    locations='State/UnionTerritory',   # The column in your df
    featureidkey='properties.ST_NM',     # The key in the GeoJSON properties
    color='Confirmed',
    color_continuous_scale="Reds",
    title="COVID-19 Case Density Across India",
    labels={'Confirmed': 'Total Cases'}
)

# This zooms the map into India specifically
fig6.update_geos(fitbounds="locations", visible=False)

fig6.show()
```
