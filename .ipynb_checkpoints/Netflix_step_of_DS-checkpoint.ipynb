**There are 6 Steps of Data Science**

1. Defining the Goal
2. Retrieving the Data
3. Data Preparation and Cleaning
4. Data Exploration
5. Data Modelling
6. Data Presentation and Automation

**Step 1- Defining the Goal**

This is the first and most important step in the data science process. In this stage, we clearly understand what problem we are trying to solve. The goal can be business-related, research-based, or prediction-focused.

Example:

In a Netflix dataset, the goal could be:
1. To analyze content trends
2. To predict movie popularity
3. To understand user preferences

**Step 2- Retrieving the Data**

In this step, data is collected from different sources such as databases, APIs, websites, or datasets (like Kaggle). The quality and relevance of data are very important for accurate results.

Data can be:
1. Structured (tables, CSV files)
2. Unstructured (text, images, videos)

Example:
Downloading the Netflix dataset from Kaggle in CSV format.

```python
#import dataset
import pandas as pd

# Load dataset
df = pd.read_csv('netflix_titles.csv')

# Show data
df.head()
```

**Step 3- Data Preparation and Cleaning**

Raw data is often incomplete, inconsistent, or contains errors. In this step, we clean and prepare the data to make it suitable for analysis.

This includes:
1. Handling missing values
2. Removing duplicates
3. Fixing incorrect data types
4. Formatting columns

**Example:** Filling missing values in the "director" column or removing duplicate Netflix entries.

```python
# Check missing values
print(df.isnull().sum())

# Fill missing values
df.fillna('Unknown', inplace=True)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Fix duration column (important step)
#df['duration_num'] = df['duration'].str.extract('(\d+)')

# Convert into number
#df['duration_num'] = pd.to_numeric(df['duration_num'], errors='coerce')

# Remove missing values for modeling
#df_model = df[['duration_num', 'release_year']].dropna()

print("Data Cleaning Done")
```

**Step 4- Data Exploration**

This step involves analyzing the dataset to understand patterns, trends, and relationships between variables. It is also called Exploratory Data Analysis (EDA).

Techniques used:
1. Data visualization (charts, graphs)
2. Identifying patterns and outliers

Example:
1. Number of movies vs TV shows
2. Content added per year
3. Top countries producing content

```python
import matplotlib.pyplot as plt

# Count Movies vs TV Shows
df['type'].value_counts().plot(kind='bar')
plt.title("Movies vs TV Shows")
plt.show()

# Content added per year
df['release_year'].value_counts().sort_index().plot(kind='line')
plt.title("Content Added Over Years")
plt.show()

# Top 5 countries
df['country'].value_counts().head().plot(kind='bar')
plt.title("Top Countries")
plt.show()
```

**Step 5- Data Modelling**

In this step, machine learning algorithms are applied to make predictions or decisions based on the data.

**Types of models:**
1. Supervised Learning (Linear Regression, Decision Tree)
2. Unsupervised Learning (Clustering)

The model is trained using data and then tested to evaluate its performance.

**Example:** Using Linear Regression to predict release year based on duration.

```python
# Convert duration column to string
df['duration'] = df['duration'].astype(str)

# Now extract numbers
df['duration_num'] = df['duration'].str.extract(r'(\d+)')

# Convert to numeric
df['duration_num'] = pd.to_numeric(df['duration_num'], errors='coerce')

# Drop missing values
df_model = df[['duration_num', 'release_year']].dropna()

# Define X and y
X = df_model[['duration_num']]
y = df_model['release_year']

# Split data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Train model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

print("Model Trained Successfully")
```

**Step 6- Data Presentation and Automation**

The final step is to present the results in a meaningful way so that others can understand and use them. This includes visualization and reporting.

Methods:
1. Graphs and dashboards
2. Reports (PDF, PowerPoint)
3. Automated scripts

Example:
Creating charts to show Netflix trends and automating analysis using Python.

```python
# Plot prediction result
plt.scatter(X_test, y_test)
plt.scatter(X_test, y_pred)
plt.title("Linear Regression Result")
plt.xlabel("Duration")
plt.ylabel("Release Year")
plt.show()
```

```python
import seaborn as sns

# Select only numeric columns
data = df[['duration_num', 'release_year']].dropna()

# Create heatmap
sns.heatmap(data.corr(), annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

The data science process is a structured approach that transforms raw data into meaningful insights. Each step plays a crucial role in ensuring accurate analysis and effective decision-making.
